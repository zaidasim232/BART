## Compute the straight line distance between every station
### Yönetm
* İlk önce konum verilerden Boylam ve Enlem kordinatları alınır.
* Haversine Förmülü uygulanarak uzaklığı hesaplanır.
#### Haversine Förmülü nedir?
1. Enlem ve Boylam Farkını Bulur: İki istasyon arasındaki yatay (boylam) ve dikey (enlem) uzaklık farkını alır.
2. Dünyanın Eğikliğini Hesaba Katar: Dünya yuvarlak olduğu için kutuplara doğru gidildikçe boylamlar birbirine yaklaşır.
3. Kuş Uçuşu Yayı Hesaplar: İki nokta arasına düz bir cetvel koymak yerine, dünyanın yüzeyi boyunca bükülen eğri yayı hesaplar.
4. Dünyanın Yarıçapı ile Çarpar: Son adımı Dünyanın yarıçapı olan $3956\text{ mil}$ (veya $6371\text{ km}$) ile çarparak sonucu bize gerçek uzunluk birimi olarak verir.
* **Özetle:** İki istasyon arasına yerin altından tünel kazmak yerine, dünya yüzeyi üstünden yürüseydik kaç mil yol giderdik, Haversine tam olarak bunu hesaplar.

```python
df_main["Origin_Longitude"] = df_main["Location_Origin"].str.split(",").str[0].astype(float)
```

```python
df_main["Origin_Longitude"].values
```

```python
df_main["Origin_Latitude"] = df_main["Location_Origin"].str.split(",").str[1].astype(float)
```

```python
df_main["Origin_Latitude"].values
```

```python
df_main["Destination_Longitude"] = df_main["Location_Destination"].str.split(",").str[0].astype(float)
```

```python
df_main["Destination_Longitude"].values
```

```python
df_main["Destination_Latitude"] = df_main["Location_Destination"].str.split(",").str[1].astype(float)
```

```python
df_main["Destination_Latitude"].values
```

* Haversine förmüle uygulamak için dereceleri radyana çevirilir.

```python
long_or = np.radians(df_main["Origin_Longitude"])
long_dest = np.radians(df_main["Destination_Longitude"])
lat_or = np.radians(df_main["Origin_Latitude"])
lat_dest = np.radians(df_main["Destination_Latitude"])
```

##### Haversine Förmülü
1. İki istasyon arasındaki yatay (boylam) ve dikey (enlem) uzaklık farkını alır.

```python
dlat = lat_dest - lat_or
dlon = long_dest - long_or
```

2. Dünyanın eğikliğini hesablanır. Formülün içindeki cos(lat1) * cos(lat2) kısmı, bulunduğun enleme göre bu daralmayı düzeltir.

```python
a = np.sin(dlat / 2) ** 2 + np.cos(lat_or) * np.cos(lat_dest) * np.sin(dlon / 2) ** 2
```

4. Dünyanın Yarıçapı ile Çarpmak.

```python
df_main["İstasyon_Uzakligi"] = 3956 * (2* np.arcsin(np.sqrt(a)))
```

```python
df_main[["Origin", "Destination", "İstasyon_Uzakligi"]].drop_duplicates().sort_values(by="İstasyon_Uzakligi", ascending=False).head(10)
```

### Feature Engineering

```python
# Weekday'dan Cumartesi ve Pazar günlerini çıkarmak için filtreleme yapıyoruz.
df_weekday = df_main[~df_main["Week_day"].isin(["Saturday", "Sunday"])]
```

```python
desc_lower = df_station["Description"].str.lower()
```

```python
# istasyonların şehir merkezi veya iş yeri olup olmadığını belirlemek için bir sütun ekliyoruz.
df_station["is_yeri"] = desc_lower.str.contains(r"\b(downtown|center|business|financial|city hall|offices)\b", regex=True).fillna(False).astype(int)
```

```python
# iş yerine yakın olan istasyonlar
df_station[df_station["is_yeri"] == 1]["Abbreviation"].unique()
```

```python
df_weekday = df_weekday.merge(df_station[["Abbreviation", "is_yeri"]].rename(
    columns={
        "Abbreviation": "Origin",
        "is_yeri": "is_yeri_Origin",
        },

), on="Origin", how="left")
```

```python
df_weekday = df_weekday.merge(df_station[["Abbreviation", "is_yeri"]].rename(
    columns={
        "Abbreviation": "Destination",
        "is_yeri": "is_yeri_Destination",
        },

), on="Destination", how="left")
```

```python
df_weekday.head()
```

* geçen haftanın yolcularını bulmak.

```python
df_weekday = df_weekday.sort_values("DateTime")
```

```python
## 168  saat 1 hafataya tekabül ettiği için shift(168) yapıyoruz.
df_weekday["1_hafta_onceki_yolcu"] = df_weekday.groupby(["Origin","Destination"])["Throughput"].shift(168)
```

```python
df_model = df_weekday[["1_hafta_onceki_yolcu","Week_day","Saat","is_yeri_Origin","is_yeri_Destination","Throughput","Origin","Destination"]]
```

### EDA

```python
df_model["is_yeri_Origin"] = df_model["is_yeri_Origin"].fillna(0)
df_model["is_yeri_Destination"] = df_model["is_yeri_Destination"].fillna(0)
```

### Regrasyon Modeli Geliştirme

```python
x = pd.get_dummies(df_model[["1_hafta_onceki_yolcu","Week_day","Saat","is_yeri_Origin","is_yeri_Destination","Origin","Destination"]])
y = df_model["Throughput"]
```

```python
#pip install xgboost
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from xgboost import XGBRegressor
from sklearn.metrics import r2_score, mean_squared_error
```

```python
x_train, x_test, y_train, y_test=train_test_split(x,y,test_size=.20,random_state=42)

scaler = StandardScaler()

x_train_scaled = scaler.fit_transform(x_train)

x_test_scaled = scaler.transform(x_test)
```

```python
xgb=XGBRegressor()
model = xgb.fit(x_train_scaled, y_train)
tahmin = model.predict(x_test_scaled)
```

```python
r2_score(y_test,tahmin)
```

```python
mean_squared_error(y_test,tahmin)**.5
```
